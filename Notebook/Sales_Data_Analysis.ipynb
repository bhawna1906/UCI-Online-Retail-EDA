## **Data Cleaning :-**
---
---

```python
df.isna().sum()
```

```python
# Removing duplicates from the dataset and saving it to a new dataframe
df_clean=df.drop_duplicates()
```

```python
# Checking the number of duplicates in the new dataframe
df_clean.duplicated().sum()
```

```python
# Filling the missing values in the 'CustomerID' column with 0 and converting the column to integer type
df_clean['CustomerID']=df_clean['CustomerID'].fillna(0).astype(int)
```

```python
# Filling the missing values in the 'Description' column with 'No Description'
df_clean['Description']=df_clean['Description'].fillna('No Description')
```

## **Feature Engineering :-**
---
---

```python
df_clean['Revenue'] = df_clean['Quantity'] * df_clean['UnitPrice']
```

```python
df_clean['Year'] = df_clean['InvoiceDate'].dt.year
```

```python
df_clean['Month']=df_clean['InvoiceDate'].dt.month
```

```python
df_clean['Month_Name'] = df_clean['InvoiceDate'].dt.month_name()
```

```python
df_clean['Quarter'] = df_clean['InvoiceDate'].dt.quarter
```

```python
df_clean['Day_Of_Week'] = df_clean['InvoiceDate'].dt.day_name()
```

```python
df_clean['Hour'] = df_clean['InvoiceDate'].dt.hour
```

```python
df_clean['Is_Cancelled'] = df_clean['InvoiceNo'].astype(str).str.startswith('C')
```

```python
df_clean[['Quantity','UnitPrice']].describe()
```

## **Metrics :-**
---
---

```python
sales = df_clean[
    (~df_clean['Is_Cancelled']) &
    (df_clean['Quantity'] > 0) &
    (df_clean['UnitPrice'] > 0)
].copy()

non_cancelled = df_clean[
    ~df_clean['Is_Cancelled']
].copy()

returns_or_negative = non_cancelled[
    (non_cancelled['Quantity'] < 0) |
    (non_cancelled['Revenue'] < 0)
].copy()
```

```python
gross_revenue = sales['Revenue'].sum()

net_revenue = non_cancelled['Revenue'].sum()

total_units_sold = sales['Quantity'].sum()

total_orders = sales['InvoiceNo'].nunique()

aov = (
    gross_revenue / total_orders
    if total_orders > 0 else 0
)

total_customers = sales.loc[
    sales['CustomerID'] != 0, 'CustomerID'
].nunique()

print(f"Gross sales revenue: {gross_revenue:,.2f}")
print(f"Net revenue after negative lines: {net_revenue:,.2f}")
print(f"Total units sold: {total_units_sold:,.0f}")
print(f"Number of sales invoices: {total_orders:,}")
print(f"Average order value (AOV): {aov:,.2f}")
print(f"Known customers: {total_customers:,}")
print(f"Cancellation invoice lines: {df_clean['Is_Cancelled'].sum():,}")
print(f"Negative/return lines: {len(returns_or_negative):,}")
```

#####  INSIGHT 

- The business generated approximately £10.64 million in gross sales from transactions meeting the notebook's positive-quantity, positive-price, and non-cancelled criteria. This provides a baseline for evaluating product performance, market contribution, and sales trends.
- Net revenue is approximately £10.62 million, around £22,124 lower than gross sales revenue. This indicates that negative-value transaction lines reduce the reported revenue, although the difference alone does not explain the full impact of cancellations and returns.
- The dataset records approximately 5.57 million units sold across eligible sales transactions. This indicates substantial transaction volume and provides a useful measure of product demand.
- The analysis identifies 19,960 unique sales invoices. This metric reflects transaction activity and provides a foundation for analyzing order frequency and average order value.
- The average order value is approximately £533.17, calculated by dividing gross sales revenue by the number of unique sales invoices.
- The analysis identifies 4,338 distinct customers with recorded customer IDs. This provides a starting point for understanding customer activity and building customer-level analyses.
- The dataset contains 9,251 lines associated with cancellation invoices. These indicate that cancellations are a relevant operational factor requiring further investigation.
- The analysis identifies 1,338 non-cancelled lines with negative quantities or negative revenue. These transactions may represent returns, credit adjustments, or other corrections.

```python
sales['InvoiceYear'] = (
    sales['InvoiceDate'].dt.to_period('Y'))

year_sales = (
    sales.groupby('InvoiceYear')
    .agg(
        Revenue=('Revenue', 'sum'),
        Units_Sold=('Quantity', 'sum'),
        Orders=('InvoiceNo', 'nunique')).reset_index())

year_sales['InvoiceYear'] = (year_sales['InvoiceYear'].astype(str))
```

```python
plt.figure(figsize=(12, 5))
sns.lineplot(
    data=year_sales,
    x='InvoiceYear',
    y='Revenue'
)

plt.title('Yearly Gross Sales Revenue')
plt.xlabel('Year')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig(r"D:\Study\Projects\UCI Online Retail\Images\Yearly_Gross_Sales_Revenue.png", dpi=300)
plt.show()
```

##### BUSINESS INSIGHT 

The yearly revenue trend provides an overview of changes in sales performance across the observation period. The business can use this trend to assess whether revenue is increasing or decreasing and to establish a baseline for future comparisons.

```python
sales['InvoiceYearMonth'] = (
    sales['InvoiceDate'].dt.to_period('M'))

monthly_sales = (
    sales.groupby('InvoiceYearMonth')
    .agg(
        Revenue=('Revenue', 'sum'),
        Units_Sold=('Quantity', 'sum'),
        Orders=('InvoiceNo', 'nunique')).reset_index())

monthly_sales['InvoiceYearMonth'] = (
    monthly_sales['InvoiceYearMonth'].astype(str))
```

```python
plt.figure(figsize=(12, 5))
sns.lineplot(
    data=monthly_sales,
    x='InvoiceYearMonth',
    y='Revenue',
    marker='o'
)

plt.title('Monthly Gross Sales Revenue')
plt.xlabel('Year-Month')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig(r"D:\Study\Projects\UCI Online Retail\Images\Monthly_Gross_Sales_Revenue.png", dpi=300)    
plt.show()
```

##### BUSINESS INSIGHT 

The monthly revenue trend helps identify fluctuations in sales activity, including potential seasonal peaks and weaker trading periods. Months with unusually high revenue can be examined to determine whether the increase came from more orders, higher order values, or changes in product demand.

```python
sales['DayOfWeek'] = sales['InvoiceDate'].dt.day_name()
day_order = [
    'Monday','Tuesday','Wednesday','Thursday','Friday','Saturday','Sunday',]

daily_sales = (
    sales.groupby('DayOfWeek')
    .agg(
        Revenue=('Revenue', 'sum'),
        Units_Sold=('Quantity', 'sum'),
        Orders=('InvoiceNo', 'nunique'),
    )
    .reindex(day_order)
    .reset_index()
)
daily_sales
```

```python
plt.figure(figsize=(10, 5))
sns.barplot(
    data=daily_sales,
    x='DayOfWeek',
    y='Revenue',
    palette='Blues_d',
    hue='DayOfWeek',
    legend=False,
)

plt.title('Total Revenue by Day of the Week', fontsize=14, fontweight='bold')
plt.xlabel('Day of the Week', fontsize=12)
plt.ylabel('Total Revenue (£)', fontsize=12)
plt.xticks(rotation=0)
plt.tight_layout()
plt.savefig(r"D:\Study\Projects\UCI Online Retail\Images\Total_Revenue_by_Day_of_the_Week.png", dpi=300)
plt.show()
```

##### BUSINESS INSIGHT 

Thursday and Tuesday generate the highest recorded revenue among the days shown, while Sunday records substantially lower revenue. This suggests that purchasing activity and transaction value vary across the week.

## **Executive Business Summary :-**
---
---


This exploratory data analysis examines the UCI Online Retail dataset to understand sales performance, purchasing patterns, product demand, and geographical revenue distribution. The analysis covers transaction data from December 2010 to December 2011 and uses data cleaning, feature engineering, key business metrics, and visualizations to identify commercially relevant patterns.

The analyzed transactions generated approximately **£10.64 million in gross sales revenue**, with 5.57 million units sold across 19,960 sales invoices. The average order value was approximately £533.17, and 4,338 distinct customers were identified through available customer IDs. These figures provide a baseline for evaluating transaction volume, order value, and customer activity.

Geographical analysis shows that the United Kingdom is the primary market, contributing approximately £9.00 million, or 84.6% of the gross sales revenue included in the analysis. This highlights the importance of the domestic market while providing a basis for evaluating international sales opportunities.

Product analysis reveals that revenue-generating descriptions and high-volume descriptions are not necessarily the same. This distinction highlights the need to evaluate products using multiple measures, including revenue, units sold, order frequency, and profitability. Postage-related entries should also be separated from physical merchandise to improve product-level comparisons.

Temporal analysis identifies variation in revenue across months, quarters, days of the week, and hours of the day. Recorded quarterly revenue increases throughout the observed quarters of 2011, with the highest value in Q4. Hourly analysis shows strong revenue activity during the late morning and early afternoon. These patterns can help guide further investigation into inventory planning, operational scheduling, and marketing timing.

The dataset also contains cancellation invoice lines and negative or return-related transactions. These require separate analysis to understand their impact on revenue and operational performance. In addition, incomplete observation periods and missing daily values must be considered before drawing conclusions about long-term growth or recurring seasonality.

Overall, the analysis provides a useful foundation for data-driven retail decisions. The main opportunities are to improve transaction monitoring, refine product-performance analysis, investigate international markets, and develop more reliable customer and profitability metrics. Further analysis incorporating product costs, customer behaviour, and additional years of data would strengthen these findings and support more confident business decisions.

## **Strategic Recommendations :-**
---
---


**1. Strengthen the Core Market and Explore International Expansion**

The United Kingdom contributes approximately 84.6% of the analyzed gross sales revenue, making it the business's primary market. The company should focus on maintaining customer satisfaction and product availability in the UK while exploring growth opportunities in international markets such as the Netherlands, Ireland, Germany, and France. Market expansion should be evaluated based on customer demand, shipping costs, and profitability.

**2. Optimize Inventory Management**

Products with high sales volumes require effective inventory planning to avoid stockouts and excess inventory. The business should identify frequently purchased products, monitor demand patterns, and establish appropriate replenishment levels. Product revenue, sales volume, and profitability should be analyzed together before making inventory decisions.

**3. Increase Average Order Value**

The average order value is approximately £533.17. The business can explore product bundling, complementary-product recommendations, and volume-based discounts to encourage customers to purchase more per order. These strategies should be evaluated to ensure that higher order values also contribute to improved profitability.

**4. Improve Sales and Operational Planning**

The analysis identifies stronger recorded sales activity during certain hours and variations in revenue across months and days of the week. The business should investigate these patterns to improve inventory availability, order processing, and operational planning. Additional years of data should be analyzed before drawing conclusions about recurring seasonal trends.

**5. Monitor Cancellations and Returns**

The dataset contains cancellation invoice lines and negative or return-related transactions. The business should investigate their underlying causes, identify frequently affected products, and improve order-processing accuracy where necessary. Tracking cancellation rates, return rates, and net revenue can help improve operational efficiency and provide a clearer understanding of sales performance.

**6. Improve Customer Retention**

The business should analyze customer purchasing histories to identify repeat customers, high-value buyers, and customers who have become inactive. Personalized offers, relevant product recommendations, and customer retention campaigns can be tested to encourage repeat purchases and strengthen long-term customer relationships.

**7. Improve Product Performance Analysis**

Products should be evaluated using multiple metrics, including revenue, units sold, order frequency, and profitability. Postage and other non-merchandise descriptions should be separated from physical products to ensure meaningful comparisons. This approach will help the business identify products that generate significant sales and determine which deserve further investment.

**8. Adopt Data-Driven Decision-Making**

The business should regularly monitor key performance indicators such as gross revenue, net revenue, average order value, sales volume, cancellation rate, and revenue by country. A sales dashboard can help management track performance, identify unusual changes, and make informed decisions based on reliable data.
