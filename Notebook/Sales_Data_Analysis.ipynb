## **Data Cleaning :-**
---
---

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
# Removing duplicates from the dataset and saving it to a new dataframe
df_clean=df.drop_duplicates()
```

```python
# Checking the number of duplicates in the new dataframe
df_clean.duplicated().sum()
```

```python
# Filling the missing values in the 'CustomerID' column with 0 and converting the column to integer type
df_clean['CustomerID']=df_clean['CustomerID'].fillna(0).astype(int)
```

```python
# Filling the missing values in the 'Description' column with 'No Description'
df_clean['Description']=df_clean['Description'].fillna('No Description')
```

```python
# Checking the number of missing values in the new dataframe
df_clean.isnull().sum()
```

---
---
## **Feature Engineering :-**
---
---

```python
df_clean['Revenue'] = df_clean['Quantity'] * df_clean['UnitPrice']
```

```python
df_clean['Year'] = df_clean['InvoiceDate'].dt.year
```

```python
df_clean['Month']=df_clean['InvoiceDate'].dt.month
```

```python
df_clean['Month_Name'] = df_clean['InvoiceDate'].dt.month_name()
```

```python
df_clean['Quarter'] = df_clean['InvoiceDate'].dt.quarter
```

```python
df_clean['Day_Of_Week'] = df_clean['InvoiceDate'].dt.day_name()
```

```python
df_clean['Hour'] = df_clean['InvoiceDate'].dt.hour
```

```python
df_clean['Is_Cancelled'] = df_clean['InvoiceNo'].astype(str).str.startswith('C')
```

```python
df_clean[['Quantity','UnitPrice']].describe()
```

---
---
## **Business Metrics :-**
---
---

```python
sales = df_clean[
    (~df_clean['Is_Cancelled']) &
    (df_clean['Quantity'] > 0) &
    (df_clean['UnitPrice'] > 0)
].copy()

non_cancelled = df_clean[
    ~df_clean['Is_Cancelled']
].copy()

returns_or_negative = non_cancelled[
    (non_cancelled['Quantity'] < 0) |
    (non_cancelled['Revenue'] < 0)
].copy()
```

```python
gross_revenue = sales['Revenue'].sum()

net_revenue = non_cancelled['Revenue'].sum()

total_units_sold = sales['Quantity'].sum()

total_orders = sales['InvoiceNo'].nunique()

aov = (
    gross_revenue / total_orders
    if total_orders > 0 else 0
)

total_customers = sales.loc[
    sales['CustomerID'] != 0, 'CustomerID'
].nunique()

print(f"Gross sales revenue: {gross_revenue:,.2f}")
print(f"Net revenue after negative lines: {net_revenue:,.2f}")
print(f"Total units sold: {total_units_sold:,.0f}")
print(f"Number of sales invoices: {total_orders:,}")
print(f"Average order value (AOV): {aov:,.2f}")
print(f"Known customers: {total_customers:,}")
print(f"Cancellation invoice lines: {df_clean['Is_Cancelled'].sum():,}")
print(f"Negative/return lines: {len(returns_or_negative):,}")
```

## **Exploratory Data Analysis With Visualization :-**

```python
plt.figure(figsize=(10,6))
sns.histplot(df_clean['Quantity'], color='blue', bins=20)
plt.title('Distribution of Quantity', fontsize=12)
plt.xlabel('Quantity', fontsize=10)
plt.ylabel('Frequency', fontsize=10)
plt.show()
```

##### BUSINESS INSIGHT 

```python
plt.figure(figsize=(10,6))
sns.histplot(df_clean['UnitPrice'], color='blue', bins=20)
plt.title('Distribution of Unit Price', fontsize=12)
plt.xlabel('Unit Price', fontsize=10)
plt.ylabel('Frequency', fontsize=10)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.histplot(df_clean['Revenue'], color='blue', bins=20)
plt.title('Distribution of Revenue', fontsize=12)
plt.xlabel('Revenue', fontsize=10)
plt.ylabel('Frequency', fontsize=10)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x=df_clean['UnitPrice'], y=df_clean['Quantity'])
plt.title('Relationship between Unit Price and Quantity', fontsize=12)
plt.xlabel('Unit Price', fontsize=10)
plt.ylabel('Quantity', fontsize=10)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x=df_clean['UnitPrice'], y=df_clean['Quantity'],hue=df_clean['Revenue'], palette='viridis', alpha=0.7)
plt.title('Relationship between Unit Price and Quantity', fontsize=12)
plt.xlabel('Unit Price', fontsize=10)
plt.ylabel('Quantity', fontsize=10)
plt.show()
```

```python
sales['InvoiceYear'] = (
    sales['InvoiceDate'].dt.to_period('Y')
)

monthly_sales = (
    sales.groupby('InvoiceYear')
    .agg(
        Revenue=('Revenue', 'sum'),
        Units_Sold=('Quantity', 'sum'),
        Orders=('InvoiceNo', 'nunique')
    )
    .reset_index()
)

monthly_sales['InvoiceYear'] = (
    monthly_sales['InvoiceYear'].astype(str)
)

plt.figure(figsize=(12, 5))
sns.lineplot(
    data=monthly_sales,
    x='InvoiceYear',
    y='Revenue',
    marker='o'
)

plt.title('Monthly Gross Sales Revenue')
plt.xlabel('Year')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
sales['InvoiceYearMonth'] = (
    sales['InvoiceDate'].dt.to_period('M')
)

monthly_sales = (
    sales.groupby('InvoiceYearMonth')
    .agg(
        Revenue=('Revenue', 'sum'),
        Units_Sold=('Quantity', 'sum'),
        Orders=('InvoiceNo', 'nunique')
    )
    .reset_index()
)

monthly_sales['InvoiceYearMonth'] = (
    monthly_sales['InvoiceYearMonth'].astype(str)
)

plt.figure(figsize=(12, 5))
sns.lineplot(
    data=monthly_sales,
    x='InvoiceYearMonth',
    y='Revenue',
    marker='o'
)

plt.title('Monthly Gross Sales Revenue')
plt.xlabel('Year-Month')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(15,6))
sns.lineplot(x=df_clean['Day_Of_Week'], y=df_clean['Revenue'],color='red')
plt.title('Revenue Trend Over the Weeks', fontsize=12)
plt.xlabel('Day of the Week', fontsize=10)
plt.ylabel('Revenue', fontsize=10)
plt.show()
```

```python
des_10=df_clean['Description'].value_counts().head(10)
des_10
```

```python
plt.figure(figsize=(10,6))
sns.barplot(x=des_10.index, y=des_10.values, color='red')
plt.title('Top 10 Product Descriptions', fontsize=12)
plt.xlabel('Description', fontsize=10)
plt.ylabel('Value', fontsize=10)
plt.xticks(rotation=45, ha='right')
plt.show()
```

```python
sales['Year'] = sales['InvoiceDate'].dt.year
sales['Quarter'] = sales['InvoiceDate'].dt.quarter

quarterly_sales = (
    sales.groupby(['Year', 'Quarter'])
    .agg(
        Revenue=('Revenue', 'sum'),
        Units_Sold=('Quantity', 'sum'),
        Orders=('InvoiceNo', 'nunique')
    )
    .reset_index()
)

print(quarterly_sales)
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    data=quarterly_sales,
    x='Quarter',
    y='Revenue',
    hue='Year',
    errorbar=None
)

plt.title('Quarterly Gross Sales Revenue by Year')
plt.xlabel('Quarter')
plt.ylabel('Revenue')
plt.tight_layout()
plt.show()
```

```python
country_sales = (
    sales.groupby('Country')
    .agg(
        Revenue=('Revenue', 'sum'),
        Units_Sold=('Quantity', 'sum'),
        Orders=('InvoiceNo', 'nunique')
    )
    .sort_values('Revenue', ascending=False)
)

print(country_sales.head(10))
```

```python
top_countries = country_sales.head(10).sort_values('Revenue')

plt.figure(figsize=(10, 6))
top_countries['Revenue'].plot(kind='barh')

plt.title('Top 10 Countries by Gross Sales Revenue')
plt.xlabel('Revenue')
plt.ylabel('Country')
plt.tight_layout()
plt.show()
```

```python
product_sales = (
    sales.groupby('Description')
    .agg(
        Revenue=('Revenue', 'sum'),
        Units_Sold=('Quantity', 'sum'),
        Orders=('InvoiceNo', 'nunique')
    )
    .sort_values('Revenue', ascending=False)
)

print("Top 10 by revenue:")
print(product_sales.head(10))

print("\nTop 10 by units sold:")
print(product_sales.sort_values(
    'Units_Sold', ascending=False
).head(10))
```

```python
top_description_by_revenue = product_sales.head(10).sort_values('Revenue')

plt.figure(figsize=(10, 6))
top_description_by_revenue['Revenue'].plot(kind='barh')

plt.title('Top 10 Products by Gross Sales Revenue')
plt.xlabel('Revenue')
plt.ylabel('Product Description')
plt.tight_layout()
plt.show()
```

```python
top_description_by_units = product_sales.head(10).sort_values('Units_Sold', ascending=False)

plt.figure(figsize=(10, 6))
top_description_by_units['Units_Sold'].plot(kind='barh')

plt.title('Top 10 Products by Units Sold')
plt.xlabel('Units Sold')
plt.ylabel('Product Description')
plt.tight_layout()
plt.show()
```
