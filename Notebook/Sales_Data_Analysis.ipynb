## **Data Cleaning :-**
---
---

```python
df.isna().sum()
```

```python
df.duplicated().sum()
```

```python
# Removing duplicates from the dataset and saving it to a new dataframe
df_clean=df.drop_duplicates()
```

```python
# Checking the number of duplicates in the new dataframe
df_clean.duplicated().sum()
```

```python
# Filling the missing values in the 'CustomerID' column with 0 and converting the column to integer type
df_clean['CustomerID']=df_clean['CustomerID'].fillna(0).astype(int)
```

```python
# Filling the missing values in the 'Description' column with 'No Description'
df_clean['Description']=df_clean['Description'].fillna('No Description')
```

```python
# Checking the number of missing values in the new dataframe
df_clean.isnull().sum()
```
